```python
def filtrar_lecturas_validas(lecturas_brutas):
    lecturas_validas = []
    for valor in lecturas_brutas:
        if valor is not None and valor >= 0:
            lecturas_validas.append(float(valor))
    return lecturas_validas

def calcular_estadisticas_descriptivas(lecturas_validas):
    total_elementos = len(lecturas_validas)
    if total_elementos == 0:
        return None

    media = sum(lecturas_validas) / total_elementos
    varianza = sum((x - media) ** 2 for x in lecturas_validas) / total_elementos
    desviacion_estandar = varianza ** 0.5

    return {
        "media": media,
        "varianza": varianza,
        "desviacion_estandar": desviacion_estandar
    }

datos_xyz = [12.5, -999, 15.0, None, 18.2, 20.1, -5, 14.8]

print("Dataset bruto original:", datos_xyz)

datos_limpios = filtrar_lecturas_validas(datos_xyz)
print("Lecturas válidas filtradas:", datos_limpios)

resultado_estadistico = calcular_estadisticas_descriptivas(datos_limpios)
print("Resultado estadístico:", resultado_estadistico)
```
