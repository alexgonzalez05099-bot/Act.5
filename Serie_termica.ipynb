```python
# Serie térmica asignada
temperaturas = [34, 18, 42, 12, 29]
n = len(temperaturas)

print("Arreglo original:", temperaturas)

# Traducción manual del pseudocódigo de Selección (Selection Sort)
for i in range(n - 1):
    indice_minimo = i
    for j in range(i + 1, n):
        if temperaturas[j] < temperaturas[indice_minimo]:
            indice_minimo = j

    if indice_minimo != i:
        # Intercambio de celdas
        temp = temperaturas[i]
        temperaturas[i] = temperaturas[indice_minimo]
        temperaturas[indice_minimo] = temp

print("Arreglo ordenado:", temperaturas)
```
